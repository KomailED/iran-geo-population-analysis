# Population of Cities in Iran
This project visualizes the population of major Iranian cities 
using a scaled scatter plot on a geographic map.

## Import Libraries

```python
import matplotlib.pyplot as plt
import pandas as pd
from mpl_toolkits.basemap import Basemap 
```

## Load and Prepare Dataset

```python
citys = pd.read_excel("citys.xlsx").to_csv("citys.csv")
citys_df = pd.read_csv("citys.csv")
color_li = ["red","blue","yellow","green","black"]
```

## Visualizing Population Distribution on the Map of Iran

```python
fig = plt.figure(figsize=(12,8),dpi=100)
axes = fig.add_axes([0.2,0.2,0.8,0.8])

m = Basemap(llcrnrlon=40.,llcrnrlat=20.,urcrnrlon=70.,urcrnrlat=45.,\
            rsphere=(6378137.00,6356752.3142),\
            resolution='l',projection='merc',\
            lat_0=35.,lon_0=55.,lat_ts=20.)
m.drawcoastlines()
m.fillcontinents()
m.drawcounties()
m.drawcountries()
m.drawstates()

x , y = m(citys_df["lon"],citys_df["lat"])
m.scatter(x,y,s=citys_df["population"]*50,color=color_li,alpha=0.4)

plt.legend()
plt.title("population of citys in Iran")
plt.xlabel("lon")
plt.ylabel("lat")
plt.show()
```
